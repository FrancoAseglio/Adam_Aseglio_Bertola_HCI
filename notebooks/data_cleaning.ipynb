## Cleaning Details


First of all let's watch more precisely which field have NaN values and **which haven't**, that wasn't explicitly shown in the **quick_profile**.

```python
details_df.isna().sum()
```

We can notice that all critical identifier and metadata columns (`mal_id`, `title`, `url`, `status`) contain no missing values, ensuring the integrity and traceability of each record.

Moreover, as we've already seen in the `data_peek` notebook other than NaN values there are even empty values such as an empty list in some field of this dataframe. Let's now focus on inspecting the empty list values.

```python
mask = details_df.astype(str).eq("[]").any(axis=1)
details_df[mask]
```

It seems that every row has at least a field with an empty value. We should take a look to every field since the table doesn't show every coloumn.

```python
details_df.iloc[1]
```

Now that we saw which fields are stored as a list, we can say that also regarding empty values all the main coloumns are valid.

In conclusion, despite many columns contain missing values (NaN) or empty list-like values ("[]"), all rows are retained.

NaN values indicate information that is not available or not yet observed, while empty lists represent valid cases where no elements are associated with the anime (e.g., no listed producers, licensors, or streaming platforms).

For this reason, empty values are not converted to NaN, as doing so would incorrectly treat the absence of associated entities as missing information and would lead to a loss of semantic meaning.

Both situations carry meaningful information, for instance, an anime without a season may be a movie, an anime without a start_date may be only announced, and an anime without an end_date may still be ongoing.

There are no duplicates as shown above in the quick_profile.

## Cleaning Profiles


Let's remove the row with NaN value in the username coloumn.


```python
# Remove username NaN values
print(f"Shape before: {profiles_df.shape}")
num_rows = profiles_df.shape[0]
profiles_df = profiles_df.dropna(subset=["username"])
print(f"Shape after:  {profiles_df.shape}")
print(f"Removed {num_rows - profiles_df.shape[0]} rows with NaN values")
```

Looking at the quick profile, we can notice that many NaN-value counters show the same number (1678).  
This suggests they might belong to the same rows, with almost all fields being NaN. Let's examine these rows.


```python
profiles_df[profiles_df["watching"].isna()]
```

As expected, most rows with NaN values are incomplete.  
However, `joined` has only 1676 NaN values - two less than the other common NaN fields (1678).  
This means there are 2 rows where `joined` is populated but the other fields (watching, completed, on_hold, dropped, plan_to_watch) might still NaN.  
Let's identify these rows.


```python
profiles_df[profiles_df["watching"].isna() & ~profiles_df["joined"].isna()]
```

These 2 rows have `joined` populated but all other activity fields (watching, completed, on_hold, dropped, plan_to_watch) are NaN, as expected.  
Since user profiles without any activity data are not useful for analysis, we can safely drop all 1678 rows where these activity fields are NaN.


```python
# Remove duplicates
print(f"Shape before: {profiles_df.shape}")
num_rows = profiles_df.shape[0]
profiles_df = profiles_df.dropna(subset=["watching"])
print(f"Shape after:  {profiles_df.shape}")
print(f"Removed {num_rows - profiles_df.shape[0]} rows with NaN values")
```

There are no duplicates as shown above in the quick_profile.

## Cleaning Stats


Similar to the profiles dataset, all columns in stats show the same NaN count.  
This strongly suggests that all NaN values belong to the same rows - completely empty entries.  
Let's verify this hypothesis by examining these rows.


```python
stats_df[stats_df["score_1_votes"].isna()]
```

It seems so, but let's check it specifically.


```python
score_cols = stats_df.loc[:, "score_1_votes":"score_10_percentage"]
nan_count_in_score_cols = score_cols.isna().sum(axis=1)
all_or_nothing_scores = nan_count_in_score_cols.isin([0, score_cols.shape[1]])
print(f"I NaN sono sempre presenti in tutte le colonne score insieme: {all_or_nothing_scores.all()}")
```

Just as we expected! We can safely remove the NaN rows.


```python
# Remove username NaN value
print(f"Shape before: {stats_df.shape}")
num_rows = stats_df.shape[0]
stats_df = stats_df.dropna()
print(f"Shape after:  {stats_df.shape}")
print(f"Removed {num_rows - stats_df.shape[0]} rows with NaN values")
```
