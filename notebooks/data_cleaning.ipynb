## Cleaning Character Nicknames

There are 16 missing nicknames as shown above in the quick_profile.

We can simpy remove them since they aren't neither relevant nor useful.

```python
# Remove NaN values
print(f"Shape before: {character_nicknames_df.shape}")
num_rows = character_nicknames_df.shape[0]
character_nicknames_df = character_nicknames_df.dropna()
print(f"Shape after:  {character_nicknames_df.shape}")
print(f"Removed {num_rows - character_nicknames_df.shape[0]} rows with NaN values")
```

Regarding the duplicates we should simply remove them too using.

```python
# Remove duplicates
print(f"Shape before: {character_nicknames_df.shape}")
num_rows = character_nicknames_df.shape[0]
character_nicknames_df = character_nicknames_df.drop_duplicates()
print(f"Shape after:  {character_nicknames_df.shape}")
print(f"Removed {num_rows - character_nicknames_df.shape[0]} duplicate rows")
```

<!--Spacing cell>

## Cleaning Characters

Since many NaN values are in coloumns which are not important for the data validity (`name_kanji`, `about`), they don't need to be removed. Instead the ones in the main coloumn (`character_mal_id`) will be removed. Let's first analyze if the rows with name NaN are valid.

```python
print(characters_df[characters_df['name'].isna()])
```

They are the same in which the character_mal_is is NaN so we can remove them safely and keep the rest.

```python
# Remove  character_mal_id NaN values
print(f"Shape before: {characters_df.shape}")
num_rows = characters_df.shape[0]
characters_df = characters_df.dropna(subset=['character_mal_id'])
print(f"Shape after:  {characters_df.shape}")
print(f"Removed {num_rows - characters_df.shape[0]} rows with NaN values")
```

Let's check that they have been removed.

```python
print(characters_df[characters_df['name'].isna()])
```
