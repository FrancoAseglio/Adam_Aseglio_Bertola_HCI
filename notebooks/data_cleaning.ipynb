## Cleaning Details

```python
# Let's see the columns and the percentage of missing values
print("Columns in details_df:")
print(details_df.columns.tolist())
print(f"\nDataframe shape: {details_df.shape}")
print("\nMissing values percentage:")
missing_pct = (details_df.isna().sum() / len(details_df) * 100).sort_values(ascending=False)
print(missing_pct)
```

```python
# Check how many rows would be lost if we drop all NaN
print(f"Total rows: {len(details_df)}")
print(f"Rows with at least one NaN: {details_df.isna().any(axis=1).sum()}")
print(f"Rows with no NaN: {(~details_df.isna().any(axis=1)).sum()}")
print(f"\nIf we drop all rows with any NaN, we would keep only {(~details_df.isna().any(axis=1)).sum()} rows")
print(f"That's {(~details_df.isna().any(axis=1)).sum() / len(details_df) * 100:.2f}% of the data")
```

Since many columns have missing values but they might still be useful, we'll keep the rows. The NaN values represent missing information (e.g., anime without season might be a film, ongoing anime without an end_date, etc.) which is valid data.

So in conclusion, there's no need to remove any rows since the critical columns (`mal_id`, `title`, `status`, `url`) contain no missing values.

There are no duplicates as shown above in the quick_profile.
