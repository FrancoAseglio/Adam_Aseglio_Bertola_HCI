## Cleaning Profiles

Let's remove the row with NaN value in the username coloumn.

```python
# Remove username NaN values
print(f"Shape before: {profiles_df.shape}")
num_rows = profiles_df.shape[0]
profiles_df = profiles_df.dropna(subset=['username'])
print(f"Shape after:  {profiles_df.shape}")
print(f"Removed {num_rows - profiles_df.shape[0]} rows with NaN values")
```

Looking at the quick profile, we can notice that many NaN-value counters show the same number (1678). This suggests they might belong to the same rows, with almost all fields being NaN. Let's examine these rows.

```python
profiles_df[profiles_df['watching'].isna()]
```

As expected, most rows with NaN values are incomplete. However, `joined` has only 1676 NaN values - two less than the other common NaN fields (1678). This means there are 2 rows where `joined` is populated but the other fields (watching, completed, on_hold, dropped, plan_to_watch) might still NaN. Let's identify these rows.

```python
profiles_df[profiles_df['watching'].isna() & ~profiles_df['joined'].isna()]
```

These 2 rows have `joined` populated but all other activity fields (watching, completed, on_hold, dropped, plan_to_watch) are NaN, as expected. Since user profiles without any activity data are not useful for analysis, we can safely drop all 1678 rows where these activity fields are NaN.

```python
# Remove duplicates
print(f"Shape before: {profiles_df.shape}")
num_rows = profiles_df.shape[0]
profiles_df = profiles_df.dropna(subset=['watching'])
print(f"Shape after:  {profiles_df.shape}")
print(f"Removed {num_rows - profiles_df.shape[0]} rows with NaN values")
```

There are no duplicates as shown above in the quick_profile.

## Cleaning Stats

Similar to the profiles dataset, all columns in stats show the same NaN count. This strongly suggests that all NaN values belong to the same rows - completely empty entries. Let's verify this hypothesis by examining these rows.

```python
stats_df[stats_df['score_1_votes'].isna()]
```

It seems so, but let's check it specifically.

```python
# Seleziona solo le colonne da score_1_votes a score_10_percentage
score_cols = stats_df.loc[:, 'score_1_votes':'score_10_percentage']

# Conta quante colonne hanno NaN per ogni riga
nan_count_in_score_cols = score_cols.isna().sum(axis=1)

# Verifica se le righe hanno o 0 NaN o tutti NaN in queste colonne
all_or_nothing_scores = nan_count_in_score_cols.isin([0, score_cols.shape[1]])
print(f"I NaN sono sempre presenti in tutte le colonne score insieme: {all_or_nothing_scores.all()}")
```

Just as we expected so we should remove the NaN rows without any problem.

```python
# Remove username NaN value
print(f"Shape before: {stats_df.shape}")
num_rows = stats_df.shape[0]
stats_df = stats_df.dropna()
print(f"Shape after:  {stats_df.shape}")
print(f"Removed {num_rows - stats_df.shape[0]} rows with NaN values")
```

This is a double check that they were all in the same rows. 430 removed -> 430 was every counter.
