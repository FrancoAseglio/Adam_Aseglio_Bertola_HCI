## Statistics Overview


```python
print("\n" + "=" * 60)
print(f"{'STATISTICS OVERVIEW':^60}")
print("=" * 60)

# Dataset Scale
print("\n### DATASET SCALE ###")
print(f"Total anime titles: {details_df.shape[0]:,}")
print(f"Total user profiles: {profiles_df.shape[0]:,}")
print(f"Total user ratings: {ratings_df.shape[0]:,}")

# Content Coverage
print("\n### CONTENT COVERAGE ###")
print(f"Character appearances in anime: {character_anime_works_df.shape[0]:,}")
print(f"Unique characters: {characters_df['character_mal_id'].nunique():,}")
print(f"Character nicknames: {character_nicknames_df.shape[0]:,}")
print(f"Anime recommendations: {recommendations_df.shape[0]:,}")

# Industry Data
print("\n### INDUSTRY DATA ###")
print(f"Industry professionals: {person_details_df.shape[0]:,}")
print(f"Voice acting roles: {person_voice_works_df.shape[0]:,}")
print(f"Staff positions: {person_anime_works_df.shape[0]:,}")

# Voice acting languages
print(f"\n## Voice acting languages:")
languages = pd.DataFrame((person_voice_works_df["language"].value_counts()))
languages_df = languages.reset_index()
languages_df.columns = ["language", "count"]
print(languages_df.to_string(index=False))

# Staff positions
print(f"\n## Top staff positions:")
positions = pd.DataFrame(person_anime_works_df["position"].value_counts())
positions_df = positions.reset_index()
positions_df.columns = ["position", "count"]
print(positions_df.head(15).to_string(index=False))

# User Demographics
print("\n### USER DEMOGRAPHICS ###")
print(f"Users with location data: {profiles_df['location'].notna().sum():,}")
top_locations = profiles_df["location"].value_counts().head(15)
print(f"\nTop 15 user locations:")
top_locations_df = top_locations.reset_index()
top_locations_df.columns = ["location", "count"]
print(top_locations_df.to_string(index=False))

# User gender distribution
print(f"\nGender distribution:")
gender_dist = profiles_df["gender"].value_counts()
gender_dist_df = gender_dist.reset_index()
gender_dist_df.columns = ["gender", "count"]
print(gender_dist_df.to_string(index=False))

# Anime Metadata
print("\n### ANIME METADATA ###")
print(f"Anime types:")
anime_types = details_df["type"].value_counts()
anime_types_df = anime_types.reset_index()
anime_types_df.columns = ["type", "count"]
print(anime_types_df.to_string(index=False))

print(f"\nAnime status:")
status_dist = details_df["status"].value_counts()
status_dist_df = status_dist.reset_index()
status_dist_df.columns = ["status", "count"]
print(status_dist_df.to_string(index=False))

print(f"\nAnime with scores: {details_df['score'].notna().sum():,}")
print(f"Average anime score: {details_df['score'].mean():.2f}")
print(f"Median anime score: {details_df['score'].median():.2f}")

# Top genres (if stored as parseable data)
print(f"\nTotal anime with genre data: {details_df['genres'].notna().sum():,}")
print(f"Total anime with themes: {details_df['themes'].notna().sum():,}")
print(f"Total anime with demographics: {details_df['demographics'].notna().sum():,}")

# User Statistics
print("\n### USER VIEWING STATISTICS ###")
print(f"Users who joined (date available): {profiles_df['joined'].notna().sum():,}")
print(f"Total anime favorites: {favs_df[favs_df['fav_type'] == 'anime'].shape[0]:,}")
print(f"Total character favorites: {favs_df[favs_df['fav_type'] == 'character'].shape[0]:,}")

# Viewing status summary
print(f"\nAverage anime per user by status:")
print(f"  - Watching: {profiles_df['watching'].astype(str).str.replace(',', '').astype(float).mean():.1f}")
print(f"  - Completed: {profiles_df['completed'].astype(str).str.replace(',', '').astype(float).mean():.1f}")
print(f"  - On hold: {profiles_df['on_hold'].astype(str).str.replace(',', '').astype(float).mean():.1f}")
print(f"  - Dropped: {profiles_df['dropped'].astype(str).str.replace(',', '').astype(float).mean():.1f}")
print(f"  - Plan to watch: {profiles_df['plan_to_watch'].astype(str).str.replace(',', '').astype(float).mean():.1f}")

# Data Completeness
print("\n### DATA COMPLETENESS ###")
print(f"Anime with Japanese titles: {details_df['title_japanese'].notna().sum():,} ({details_df['title_japanese'].notna().sum()/len(details_df)*100:.1f}%)")
print(f"Anime with synopsis: {details_df['synopsis'].notna().sum():,} ({details_df['synopsis'].notna().sum()/len(details_df)*100:.1f}%)")
print(f"Characters with descriptions: {characters_df['about'].notna().sum():,} ({characters_df['about'].notna().sum()/len(characters_df)*100:.1f}%)")
print(f"Characters with kanji names: {characters_df['name_kanji'].notna().sum():,} ({characters_df['name_kanji'].notna().sum()/len(characters_df)*100:.1f}%)")
```
