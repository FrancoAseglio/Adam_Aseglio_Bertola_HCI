# Data Analysis: Anime

We've been tasked with analyzing the public dataset from **MyAnimeList (MAL)**, a platform where users rate anime and organize them into lists  
—"Plan to Watch," "Watching," "On Hold," "Dropped," or "Completed"—  
with date tracking for each entry.  
Beyond ratings, MAL lets users favorite characters, follow anime news, and join topic-specific clubs (essentially themed group chats for fans).

This dataset captures the entire anime ecosystem: what people watch, how they engage, and what they love.

This dataset is organized into four macro categories that together describe the anime ecosystem from content to audience.  
The Anime category covers the titles themselves and how people interact with them.  
The `details` table holds core per-title metadata such as titles, formats, genres, airing dates, and platforms.  
The `stats` table captures aggregate engagement, including how many users are watching, completed, or dropped each title, along with score distributions.  
The `ratings` table contains individual user logs that record per-user watch status and numeric ratings.  
The `recommendations` table links titles with "if you liked X, try Y" relationships that help surface similar shows.

The Characters category represents the cast and how they appear across shows.  
The `characters` table provides biographies, including identity, background, and notable traits.  
The `character_anime_works` table maps each character to the anime in which they appear and indicates whether their role is main or supporting.  
The `character_nicknames` table lists alternate names and fan-used monikers, reflecting how communities refer to characters in practice.

The Creators category documents the people who make anime happen and their credited work.  
The `person_details` table profiles industry professionals such as voice actors, directors, and animators.  
The `person_anime_works` table records production credits, specifying who worked on which anime and in what capacity.  
The `person_voice_works` table focuses on voice acting, showing who voiced which characters and in which language.  
The `person_alternate_names` table captures stage names and alternative spellings used by creators.

The Users category describes the audience, their habits, and their preferences.  
The `profiles` table summarizes user demographics and viewing behavior, including how much they have watched, completed, dropped, placed on hold, or plan to watch.  
The `favs` table highlights user favorites, identifying the anime and characters individual users love most.

Taken together, these four categories offer a coherent map of the dataset:

- titles and engagement live under Anime;
- personality and presence live under Characters;
- production roles and credits live under Creators;
- audience behavior and preferences live under Users.

## What makes a character resonate the most with audience?

### **As expected: The Protagonist's Unfair Advantage**

Supporting characters outnumber mains 4-to-1, but main characters are 8× more beloved.  
The narrative spotlight doesn't just boost popularity, it transforms engagement entirely.

**Why the gap exists:**
Main characters get backstory, arc, internal monologue, pivotal decisions. Supporting characters get reaction shots.  
You can't love what you don't know.

**The POV effect:** Anime makes you _become_ the protagonist. Their struggles become yours.  
Supporting characters remain external; observed, not inhabited.

**The arc requirement:** "Favorite" means narrative journey. Supporting characters are snapshots, not stories.  
They rarely transform because their change would distract from the protagonist.

**The uncomfortable truth:** Viewers form deep bonds with 1-2 mains while casually appreciating everyone else.  
The gap isn't a failure, it's how human attention actually works.

**What this means:** Main character quality is non-negotiable. A perfect supporting cast cannot save a weak protagonist.  
Stories need centers and protagonists are the emotional anchor, the viewer's entry point, the lens through which everything gains meaning.

### **The Character Efficiency Paradox**

This isn't about which genres are popular, it's about **character-to-emotion conversion efficiency.** How many favorites does each character generate?

**Suspense wins big:** Highest favorites with the smallest rosters. Every character must matter because the genre demands psychological depth.  
No throwaway characters in psychological thrillers.

**Award Winning = Character Memorability:** You win awards for characters that haunt viewers, not plots.  
These anime treat characters as literature, complex beings worth extended exploration.

Archetypes, not personalities.

**Mystery's weakness:** The puzzle overshadows the person. Viewers remember the twist, not always who delivered it.

**The lesson:** Small cast with deep investment beats large cast with shallow investment, but genre must support that strategy.

## The Voice Behind the Character

If language drives adoption this strongly, what about the voice actors themselves?  
Do certain VAs consistently voice beloved characters, or is popularity purely about the character and story?  
Let's explore whether talent matters as much as localization.

**Reading the visualization:**

This reveals whether top VAs got there by voicing many characters, or by consistently landing iconic roles.

- **Bar length** = Total favorites across all characters voiced by that VA

- **Bar color** = Average favorites per character (darker = higher average)

- **Text inside bars** = Number of characters voiced


```python
fig, ax = gu.voice_actor_character_popularity()
```

### **Does the voice actor matter?**

**For character popularity? Not as much as you might expect.**

The data reveals there isn't a single "formula" for success.  
Some VAs reach the top through volume (Mayumi Tanaka: 230 characters), while others achieve it with selective iconic roles (Hiroaki Iwanaga: 87k avg from just 9 characters).

The key insight: **high average favorites per character (right chart) doesn't guarantee a top total impact (left chart)**.  
You can succeed with either consistency across many roles OR a few breakout performances—but there's no dominant strategy.

Looking at this data, we see that among already-popular characters, there's no clear evidence that being voiced by a top VA guarantees success.

- Mayumi Tanaka (230 chars) has medium-low average favorites despite huge volume
- Hiroaki Iwanaga (9 chars) has the HIGHEST average but far lower total impact
- Daniel Schlauch appears in BOTH lists (63 chars, high total AND high average)—but he's the exception

**What this suggests:**  
Character popularity depends more on the character itself (design, personality, story role) than who voices them.  
VAs enhance the experience, but they don't create popularity from nothing.

**Contrast with localization:**

Unlike localization—which shows clear, direct correlation with regional user adoption—voice actors operate at the micro-level.  
Localization opens doors to entire markets; voice acting polishes what's already there.

**Takeaway for producers:**

Voice acting is **critical for execution, not for generating buzz**.

- A great VA elevates a well-written character
- A famous VA can't save a poorly-designed or narratively weak character

**Strategy:**  
Invest in quality voice acting to maximize your characters' potential,  
but don't expect VA star power alone to carry your marketing or viewership numbers.

## The Commitment Question: What Makes Us Finish What We Start?

Every anime fan knows the feeling: you queue up a new series, watch the first few episodes, and then… life happens.  
The show sits in your "Currently Watching" list for months. Maybe forever.

But what separates the anime we finish from the ones we abandon?  
Is it purely a matter of quality, or is there something deeper at play—a psychological contract between viewer and content that either holds or breaks?

This analysis dives into the completion and dropout patterns across thousands of anime titles and millions of viewing decisions.  
We examine three critical dimensions:

- **Episode Count**: Does length breed commitment or fatigue?
- **Format**: Are movies inherently more "finishable" than TV series?
- **Genre**: Do emotional stakes trump action spectacle when it comes to keeping viewers hooked?

The findings reveal a nuanced picture: **not all content is meant to be completed, and that's not necessarily a failure.**  
Some anime are designed to be lived with rather than finished. Others promise—and must deliver—closure in a single sitting.

Understanding these patterns isn't just academic.  
For creators, it's a roadmap for designing content that respects the viewer's time.  
For platforms, it's a guide to retention strategy.  
And for fans? It's validation that dropping a show doesn't make you a quitter—it makes you human.

Let's break down the data.

```python
fig = gu.completion_dropout_by_episode_count()
fig.show()
```

### **The Commitment Contract: Why We Finish (or Drop) Anim**e

Anime isn't just judged by quality; it's judged by the **time it demands versus the payoff it promises.** Every "Episode 1" is a silent negotiation.

### 1. The Browser (1–24 Episodes)

- **The Profile:** Low-risk, high-reward. (Movies, OVAs, 12-episode seasons).
- **The Data:** Completion is highest here, ranging from **70% to 79%**.
- **The Psychology:** Browsers trade a weekend for a tight narrative. They hate "filler" and are easily scared off by indefinite hiatuses.  
  A cliffhanger without a confirmed Season 2 is a broken contract.

### 2. The Investor (25–100 Episodes)

- **The Profile:** The mid-length commitment.
- **The Data:** We hit the **"Commitment Cliff"**—completion slides from **59% to 56%**.
- **The Psychology:** This is where casuals are filtered out. To keep an Investor, the story must justify its length through world-building.  
  If a show treads water or repeats arcs, the viewer realizes their time is being wasted and walks away.

### 3. The Devotee (101+ Episodes)

- **The Profile:** Long-running sagas and mega-series.
- **The Data:** Completion **stabilizes at ~50%**.
- **The Psychology:** The drop-off stops because "tourists" never start these. Those who remain aren't just viewers; they are residents.  
  The show becomes a weekly ritual—a second home.

### The 28% Rule: The Value of the "Unfinished"

For 100+ episode series, the **In-Progress rate hits 28.6%**.  
From a platform perspective, this is the holy grail. While "Completed" users might leave, "In-Progress" users stay subscribed.  
These sagas aren't meant to be finished; they are meant to be **lived with.**

### The Bottom Line: Format is a Promise

- **Movies** promise closure in 90 minutes.
- **Short Series** promise satisfaction in a weekend.
- **Long Sagas** promise a world to live in for years.

```python
fig = gu.completion_dropout_by_type()
fig.show()
```

### **Viewing Habits: Does Format Predict Whether a User Will Stick It Out?**

The data suggests a clear divide: brevity is the best friend of completion, while long-form content faces an uphill battle against viewer fatigue.

### The "Quick Win" Phenomenon

It’s no surprise that **Music Videos (86.0%)**, **Specials (81.0%)**, and **Promotional Videos (80.9%)** boast the highest completion rates.  
These formats offer high satisfaction with minimal time investment.  
Because they are often one-offs or just a few minutes long, the "barrier to entry" is almost non-existent.  
These formats also enjoy the lowest dropout rates (all under 9%), proving that when the finish line is in sight from the start, viewers are far more likely to cross it.

### The Struggle of the Long-Form Narrative

On the flip side, **TV series (64.6%)** and **Original Net Animations (ONA) (63.1%)** represent the biggest challenge for retention.  
Even though TV series make up the lion's share of the industry—with over 660 million started users and 6,000+ titles—they suffer from significant dropout rates (around 15.9%).

This suggests a few things:

- **Quality Variance:** With so many TV titles available, the "middle-of-the-pack" shows often lose viewers halfway through.
- **Viewer Fatigue:** Maintaining interest over 12, 24, or hundreds of episodes is a heavy lift for any creator.
- **The ONA Paradox:** Despite being easily accessible on web platforms, ONAs struggle the most with retention, perhaps due to their often experimental or inconsistent nature.

### The Sweet Spot: Movies and OVAs

Sitting comfortably in the middle are **Movies (75.8% completion)** and **OVAs (72.4%)**. Movies benefit from being self-contained; they offer a complete narrative arc in a single sitting, making them more "finishable" than a TV season but more substantial than a music video.

### What This Means for the Industry

There is a clear inverse correlation at play: as the length and complexity of a format increase, the likelihood of a user finishing it decreases.

- **For Creators:** Shorter formats are a safe bet for ensuring your entire message is seen.
- **For Platforms:** While TV series drive massive volume and keep users coming back to the app, they require much higher quality control or "hook" mechanics to prevent nearly 1 in 6 viewers from walking away.

### **Genre Matters: Why Some Stories Keep Us Hooked While Others Fade**

When we look beyond the format of an anime and dive into its genre, a different story emerges.

### The Power of the Heart: Romance and Drama

If you want to ensure a viewer reaches the finale, tell them a love story.  
**Romance** leads the pack with a staggering **78.5% completion rate**.  
Closely following are **Supernatural (75.6%)** and **Drama (75.0%)**.  

What do these three have in common?

They lean heavily on emotional stakes and narrative momentum.  
When a viewer becomes invested in a character's relationship or a central mystery, they feel a psychological "need" to see the resolution.  
This is reflected in the dropout rates, too—Romance sees only 7.2% of viewers walk away, the lowest of any major genre.  

### The "Casualty" of Casual Content

Interestingly, **Slice of Life** and **Comedy** struggle the most with retention.
Slice of Life, in particular, has one of the lowest completion rates at **67.5%** and the highest dropout rate at **16.1%**.

This suggests that the "episodic" nature of these genres is a double-edged sword.  
While it makes them easy to jump into for a quick laugh, the lack of an urgent, overarching plot makes it just as easy for a viewer to stop watching whenever they feel like it.  
Without a "cliffhanger" or a clear end-goal, the incentive to finish the entire series drops.  

### Volume Doesn't Always Mean Engagement

The relationship between a genre's popularity and its completion rate is eye-opening:  

- **The "Safe Bets":** **Action** is a powerhouse, maintaining a high volume (417M users) and a healthy 73.5% completion rate. It's the industry's most reliable performer.  
- **The Fantasy Fatigue:** Despite its massive popularity (273M users), **Fantasy** falls into the lower tier of completion at **69.6%**.  
  This suggests that while viewers are drawn to magical worlds, many series in this saturated market fail to sustain interest past the initial world-building phase.  
- **The Adult Niche:** **Hentai** sits at the bottom of the completion list (67.4%).  
  While its dropout rate isn't the highest, the data suggests that for adult content, viewers are often looking for specific moments rather than a complete narrative journey.

### **The Bell Curve of Success: Where Most Anime Actually Land**

The distribution tells a story that genre averages hide: **most anime cluster around mediocrity.**

**The Completion Reality:**
- The peak sits around **70-75% completion**—this is "good enough" territory
- Very few titles break into the 85%+ elite tier
- The left tail shows a significant number of anime with sub-50% completion—these are the failures, the ones viewers start but quickly abandon

**The Dropout Mirror:**
- Most anime hover around **10-15% dropout**—this is the "acceptable loss" zone
- High dropout rates (25%+) are rare but devastating when they occur
- The tightest clustering is in the 8-12% range—this is the industry standard

**What the shape reveals:**
The distribution is relatively **tight and centered**, not wildly spread out. This means:
- There's a baseline quality floor in the industry—truly terrible anime don't get made (or tracked) often
- Breaking out from the pack is incredibly difficult—marginal improvements in storytelling don't dramatically shift completion
- The "hits" aren't statistical outliers; they're incremental winners in a crowded field

**The Producer's Dilemma:**
With most content performing similarly, differentiation becomes everything.  
You're not competing against terrible anime—you're competing against hundreds of "pretty good" ones.  
The data shows that **excellence is rare, but so is catastrophic failure.**  
Most creators land in the comfortable middle, which is simultaneously safe and invisible.
